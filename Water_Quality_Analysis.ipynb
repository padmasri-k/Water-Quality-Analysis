```python
data_clean = data.copy()
```

```python
columns_to_fill = [
    'Chl a [mg/m**3]',
    'BOD [mg/l]',
    'pH',
    'DO [mg/l]',
    'TSS [mg/l]',
    'Turbidity [NTU]',
    'TI'
]

for column in columns_to_fill:
    data_clean[column] = data_clean[column].fillna(data_clean[column].median())
```

```python
data_clean.shape
```

```python
import matplotlib.pyplot as plt
```

```python
columns = [
    'pH',
    'Turbidity [NTU]',
    'DO [mg/l]',
    'TDS [mg/l]',
    'WQI'
]

for column in columns:
    plt.figure(figsize=(6, 4))
    plt.boxplot(data_clean[column].dropna())
    plt.title(f'Box Plot of {column}')
    plt.ylabel(column)
    plt.show()
```

```python
for column in columns:
    Q1 = data_clean[column].quantile(0.25)
    Q3 = data_clean[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data_clean[
        (data_clean[column] < lower_bound) |
        (data_clean[column] > upper_bound)
    ]

    print(f"{column}: {len(outliers)} outliers")
```

```python
columns = [
    'pH',
    'Turbidity [NTU]',
    'DO [mg/l]',
    'TDS [mg/l]',
    'WQI'
]

for column in columns:
    Q1 = data_clean[column].quantile(0.25)
    Q3 = data_clean[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data_clean[
        (data_clean[column] < lower_bound) |
        (data_clean[column] > upper_bound)
    ][[column]]

    print("\n", column)
    print(outliers)
```

```python
site_analysis = data_clean.groupby('Site')[
    ['pH', 'DO [mg/l]', 'TDS [mg/l]', 'Turbidity [NTU]', 'WQI']
].mean()

site_analysis
```

```python
plt.figure(figsize=(12,5))

site_analysis['WQI'].plot(kind='bar')

plt.title('Average WQI by Sampling Site')
plt.xlabel('Sampling Site')
plt.ylabel('Average WQI')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
correlation_columns = [
    'pH',
    'DO [mg/l]',
    'TDS [mg/l]',
    'Turbidity [NTU]',
    'TSS [mg/l]',
    'BOD [mg/l]',
    'TN [mg/l]',
    'TP [mg/l]',
    'EC [µS/cm]',
    'WQI'
]

correlation_matrix = data_clean[correlation_columns].corr()

correlation_matrix
```

```python
plt.figure(figsize=(10,8))

plt.imshow(correlation_matrix, cmap='coolwarm', aspect='auto')

plt.colorbar()

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=90
)

plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

plt.title('Correlation Heatmap of Water Quality Parameters')

plt.tight_layout()
plt.show()
```
